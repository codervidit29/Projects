```python
# prompt: group by season, winner and select only the max count winner from each season. order by max count

season_winner_counts = df.groupby(['season', 'winner'])['winner'].count().reset_index(name='counts')
max_winner_per_season = season_winner_counts.loc[season_winner_counts.groupby('season')['counts'].idxmax()]
max_winner_per_season = max_winner_per_season.sort_values(by='counts', ascending=False)
max_winner_per_season
```

```python
# prompt: using max_winner_per_season, group by winner and sum of counts. order by sum of counts

winner_counts = max_winner_per_season.groupby('winner')['counts'].sum().sort_values(ascending=False)
winner_counts
```

```python
# prompt: group by winner. sum of count of winner. order by sum of count of winner

# Group by winner, sum the counts, and order by the sum of counts
winner_counts = df.groupby('winner')['winner'].count().sort_values(ascending=False)
winner_counts
```

```python
# prompt: bar plot for winner_counts

import matplotlib.pyplot as plt
plt.figure(figsize=(12, 6))
sns.barplot(x=winner_counts.index, y=winner_counts.values)
plt.xlabel('Winner')
plt.ylabel('Number of Wins')
plt.title('Number of Wins per Team')
plt.xticks(rotation=90)
plt.show()
```

```python
# prompt: group by toss_winner. sum of count of toss_winnner. order by sum of count of toss_winnner decending

toss_winner_counts = df.groupby('toss_winner')['toss_winner'].count().sort_values(ascending=False)
toss_winner_counts
```

```python
# prompt: group by toss decision. sum of count of toss decision. order by sum of count of toss decision

toss_decision_counts = df.groupby('toss_decision')['toss_decision'].count().sort_values(ascending=False)
toss_decision_counts
```

```python
# prompt: pie chart for toss_decision_counts

import matplotlib.pyplot as plt
# Assuming 'toss_decision_counts' is already defined as in your previous code
# If not, uncomment the following line and replace with your actual data
# toss_decision_counts = df.groupby('toss_decision')['toss_decision'].count().sort_values(ascending=False)

# Create the pie chart
plt.figure(figsize=(8, 8))  # Adjust figure size as needed
plt.pie(toss_decision_counts.values, labels=toss_decision_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Toss Decision Distribution')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle
plt.show()
```

```python
# prompt: group by winners. count of win by runs is not equal to zero. order by count of win by runs is not equal to zero

# Filter out rows where 'win_by_runs' is not equal to zero
win_by_runs_not_zero = df[df['win_by_runs'] != 0]

# Group by 'winner' and count the occurrences where 'win_by_runs' is not zero
win_counts = win_by_runs_not_zero.groupby('winner')['win_by_runs'].count()

# Order by the counts in descending order
win_counts_ordered = win_counts.sort_values(ascending=False)

win_counts_ordered
```

```python
# prompt: group by winners. count of win by wickets is not equal to zero. order by count of win by wickets is not equal to zero

# Filter out rows where 'win_by_wickets' is not equal to zero
win_by_wickets_not_zero = df[df['win_by_wickets'] != 0]

# Group by 'winner' and count the occurrences where 'win_by_wickets' is not zero
win_counts = win_by_wickets_not_zero.groupby('winner')['win_by_wickets'].count()

# Order by the counts in descending order
win_counts_ordered = win_counts.sort_values(ascending=False)

win_counts_ordered
```
